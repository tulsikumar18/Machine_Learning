# Column Transformer (Feature Preprocessing Technique)

## What is Column Transformer?

**Column Transformer** is a **Machine Learning preprocessing technique** used to apply **multiple preprocessing techniques** to different feature columns **at the same time**.

It allows us to preprocess different columns in **parallel** and combines the transformed output into a single dataset.

> **Note:** Column Transformer is used **only for preprocessing techniques**, **not for Machine Learning algorithms**.

---

# Why do we need Column Transformer?

In a real-world dataset, different columns often require different preprocessing techniques.

Example:

| Feature | Preprocessing Required |
|---------|------------------------|
| Age | MinMaxScaler |
| Salary | RobustScaler |
| City | OneHotEncoder |
| Occupation | TargetEncoder |

Without a Column Transformer, we would need to preprocess each column separately.

With a Column Transformer, all preprocessing steps are applied together in a single pipeline.

---

# Workflow

```text
              Input Features
                    │
      ┌─────────────┼─────────────┐
      │             │             │
      ▼             ▼             ▼
 MinMaxScaler  RobustScaler  TargetEncoder
      │             │             │
      └─────────────┼─────────────┘
                    │
                    ▼
         Column Transformer
                    │
                    ▼
     Combined Transformed Dataset
```

---

# When to Use Column Transformer?

Use a Column Transformer when different feature columns require different preprocessing techniques.

Example:

- Numerical columns → Scaling
- Categorical columns → Encoding
- Text columns → Vectorization (if required)

---

# Common Preprocessing Techniques

| Technique | Best Used For |
|-----------|---------------|
| MinMaxScaler | Numerical data with **few or no outliers** |
| RobustScaler | Numerical data with **many outliers** |
| StandardScaler | Normally distributed numerical data |
| OneHotEncoder | Nominal categorical features |
| OrdinalEncoder | Ordered categorical features |
| TargetEncoder | High-cardinality categorical features |

---

# Syntax

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessing = ColumnTransformer(

    transformers=[

        ('name1', preprocessing_object, column_names),

        ('name2', preprocessing_object, column_names)

    ],

    remainder='drop'   # or 'passthrough'

)
```

---

# Example

Suppose the dataset contains:

| Feature | Preprocessing |
|---------|---------------|
| f1, f2 | MinMaxScaler |
| f3, f4 | RobustScaler |
| f5, f6 | TargetEncoder |

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, RobustScaler
from category_encoders import TargetEncoder

preprocessing = ColumnTransformer(

    transformers=[

        ('minmax_scaling', MinMaxScaler(), ['f1', 'f2']),

        ('robust_scaling', RobustScaler(), ['f3', 'f4']),

        ('target_encoding', TargetEncoder(), ['f5', 'f6'])

    ]

)
```

---

# Applying the Transformation

### Training Data

```python
X_train_new = preprocessing.fit_transform(X_train, y_train)
```

> **Note:** `TargetEncoder` requires the target column (`y_train`) during `fit_transform()`.

### Testing Data

```python
X_test_new = preprocessing.transform(X_test)
```

Only `transform()` is used because the preprocessing parameters have already been learned from the training data.

# `remainder` Parameter

The `remainder` parameter controls what happens to the columns **not mentioned** in the `transformers` list.

---

## 1. `remainder='drop'` (Default)

Columns that are **not specified** are removed.

Example

Dataset

| f1 | f2 | f3 | f4 | f5 | f6 |
|----|----|----|----|----|----|

Only these columns are transformed:

```python
['f1', 'f2', 'f3', 'f4']
```

Result

```text
Only transformed columns are returned.

f5 and f6 are dropped.
```

---

## 2. `remainder='passthrough'`

Columns that are **not specified** are kept unchanged.

```python
preprocessing = ColumnTransformer(

    transformers=[

        ('scaler', MinMaxScaler(), ['f1', 'f2'])

    ],

    remainder='passthrough'

)
```

Output

```text
Scaled Columns

+

Remaining Columns

↓

Final Dataset
```

All original columns are preserved.

---

# Advantages

- Applies multiple preprocessing techniques in one step.
- Executes preprocessing in parallel.
- Combines transformed columns into one dataset.
- Reduces repetitive preprocessing code.
- Easy to integrate with machine learning pipelines.
- Makes preprocessing more organized and maintainable.

---

# Limitations

- Used only for **feature preprocessing**.
- Cannot be used to train machine learning models.
- Requires correct column selection for each transformer.

---

# Best Practices

- Apply preprocessing **after Train-Test Split**.
- Use `fit_transform()` only on the training data.
- Use `transform()` on testing or new data.
- Choose the preprocessing technique based on the type of feature.
- Use `TargetEncoder` only when high-cardinality categorical features are present.

# Decision Tree

A **Decision Tree** is a **Supervised Machine Learning algorithm** used for both **Classification** and **Regression**. It makes predictions by asking a **series of decision-based questions**, where each answer splits the dataset into smaller groups until a final prediction (leaf node) is reached.

> **Idea:** The tree keeps splitting the data until the nodes become as **pure** as possible.

---

# Applications

- Email Spam Detection
- Disease Prediction
- Loan Approval
- Customer Churn Prediction
- House Price Prediction (Regression)

---

# Important Terms

### 1. Root Node
- The **first (top-most)** node of the tree.
- Represents the **best feature** selected for the first split.

---

### 2. Internal Node

- Intermediate nodes that contain decision questions.
- An internal node can further split into child nodes.

> Internal nodes are generally **Impure Nodes**.

---

### 3. Leaf Node

- Final node of the tree.
- Gives the final prediction/output.
- Cannot be split further.

> Leaf nodes are generally **Pure Nodes**.

---

### 4. Split

A **Split** is the process of dividing a node into child nodes based on a feature.

> Decision Trees generally perform **Binary Splits** (2 branches).

---

### 5. Purity

A node containing samples of **only one class**.

Example

| Email | Label |
|-------|-------|
| Spam | Spam |
| Spam | Spam |
| Spam | Spam |

This is a **Pure Node** because every sample belongs to the same class.

---

### 6. Impurity

A node containing **multiple class labels**.

Example

| Email | Label |
|-------|-------|
| Spam | Spam |
| Offer | Spam |
| Meeting | Not Spam |

This is an **Impure Node** because both classes are present.

---

# Relationship

```text
Internal Node
      │
      ▼
 Impure Node
      │
 Split Again
      │
      ▼
 Pure Node
      │
      ▼
 Leaf Node
```

# How Does a Decision Tree Select the Best Feature?

Decision Trees select the **best feature** using the **Criterion** parameter.

The feature with the **lowest impurity (or highest information gain)** becomes the **Root Node** or an **Internal Node**.

---

# Criterion

Criterion is a measure used to determine **which feature creates the best split**.

There are two commonly used criteria:

### 1. Gini Impurity (Default)

- Based on **probability**.
- Faster to compute.
- Default criterion in Scikit-Learn.

```python
criterion="gini"
```

---

### 2. Entropy

- Based on **Information Theory**.
- Uses **log₂**.
- Measures the amount of uncertainty in the data.

```python
criterion="entropy"
```

---

## Comparison

| Gini | Entropy |
|------|----------|
| Uses Probability | Uses log₂ |
| Faster | Slightly Slower |
| Default in Scikit-Learn | Optional |
| Measures Impurity | Measures Information Gain |

---

# Splitter Parameter

The **Splitter** determines **how candidate split points are selected**.

There are two options:

1. **best** (Default)
2. **random**

---

## 1. Splitter = "best"

The algorithm evaluates **all possible split points** and chooses the one that gives the **best criterion value**.

Example

```

15
17
8
5
5
```

```

```

### Step 2

```

```

### Step 3

```

```

### Step 4

Find midpoint of each range

```

```

### Step 5

For every midpoint,

the Decision Tree calculates the **Criterion (Gini/Entropy)**.

The split having the **lowest impurity** (or highest information gain) is selected as the **Best Split**.

# Important Parameters of Decision Tree

Both **DecisionTreeClassifier** and **DecisionTreeRegressor** have several parameters that control how the tree grows and help prevent **overfitting**.

---

# 1. max_depth

```python
max_depth = None | int
```

- Controls the **maximum depth (height)** of the decision tree.
- By default (`None`), the tree grows until all leaf nodes are pure (Classifier) or other stopping conditions are met.
- A smaller value helps reduce **overfitting**.

### Example

```python
DecisionTreeClassifier(max_depth=5)
```

The tree can grow only up to **5 levels**.

---

# 2. min_samples_split

```python
min_samples_split = 2 | int
```

- Minimum number of samples required to **split an internal node**.
- If a node contains fewer samples than this value, it **cannot be split further**.

### Example

```python
min_samples_split = 10
```

| Samples at Node | Split Allowed? |
|-----------------|----------------|
| 20 | ✅ Yes |
| 10 | ✅ Yes |
| 8 | ❌ No |

---

# 3. min_samples_leaf

```python
min_samples_leaf = 1 | int
```

- Minimum number of samples that **must remain in every leaf node** after splitting.
- Prevents the tree from creating very small leaf nodes.

### Example

```python
min_samples_leaf = 5
```

A split is allowed only if **both child nodes contain at least 5 samples**.

---

# Example of `min_samples_split` and `min_samples_leaf`

Suppose:

```text
Total Samples = 1000

min_samples_split = 10
min_samples_leaf = 101
```

Decision Tree

```text
                    Root (1000)
                         │
                 Split Allowed ✔
                    (1000 ≥ 10)
                         │
               ┌─────────┴─────────┐
               │                   │
           Node A (200)        Node B (800)
               │                   │
          Split Allowed ✔     Split Allowed ✔
          (200 ≥ 10)          (800 ≥ 10)

               │
         ┌─────┴─────┐
         │           │
      50 Samples   150 Samples

```

Here,

- Left child = **50 samples**
- Right child = **150 samples**

Since

```text
50 < 101
```

the left child violates `min_samples_leaf`.

Therefore,

**the entire split is rejected**, even though the right child satisfies the condition.

The node remains a **leaf node**.

> **Rule:** A split is valid only if **both child nodes satisfy `min_samples_leaf`.**

---

# 4. max_features

```python
max_features = float | "sqrt" | "log2" | None
```

Controls the **number of features considered while finding the best split**.

### Options

### None (Default)

Uses **all features**.

```python
max_features=None
```

---

### Float

A fraction of total features.

```python
max_features=1.0
```

Uses **100% of the features**.

Example:

```
20 Features

↓

1.0 × 20 = 20 Features
```

---

### sqrt

Uses

```text
√(Total Features)
```

Example

```
Total Features = 144

√144 = 12
```

Randomly selects **12 features** at each split.

---

### log2

Uses

```text
log₂(Total Features)
```

Example

```
16 Features

↓

log₂16 = 4 Features
```

---

# 5. random_state

```python
random_state = int | None
```

- Controls randomness in the algorithm.
- Produces **reproducible results**.

Example

```python
DecisionTreeClassifier(random_state=42)
```

Using the same random state gives the same tree every time.

---

# 6. class_weight (Classifier Only)

```python
class_weight = "balanced" | None
```

Used only in **DecisionTreeClassifier**.

Helps handle **imbalanced datasets** by assigning higher importance to minority classes.

Example

```python
DecisionTreeClassifier(class_weight="balanced")
```

> **Note:** `class_weight` is **not available** in `DecisionTreeRegressor`.

---

# Difference Between DecisionTreeClassifier and DecisionTreeRegressor

| Parameter | DecisionTreeClassifier | DecisionTreeRegressor |
|-----------|-------------------------|------------------------|
| criterion | `gini`, `entropy`, `log_loss` | `squared_error`, `absolute_error`, `friedman_mse`, `poisson` |
| class_weight | ✅ Available | ❌ Not Available |

All other important parameters (`max_depth`, `min_samples_split`, `min_samples_leaf`, `max_features`, `random_state`) are common to both.

---

# Criterion in DecisionTreeRegressor

The regressor measures the quality of a split using prediction error.

### 1. squared_error (Default)

Uses **Mean Squared Error (MSE)**.

Formula

```text
                 n
MSE = (1/n) × Σ (yi − ȳ)²
                i=1
```

Where:

- `yi` = Actual value
- `ȳ` = Mean of target values
- `n` = Number of samples

---

### 2. absolute_error

Uses **Mean Absolute Error (MAE)**.

Formula

```text
                 n
MAE = (1/n) × Σ |yi − ȳ|
                i=1
```

MAE is less sensitive to outliers than MSE.

# Stopping Condition

### DecisionTreeClassifier

Stops splitting when:

- A **Pure Node** is obtained (all samples belong to one class), or
- Stopping parameters (`max_depth`, `min_samples_split`, `min_samples_leaf`) are reached.

Example

```text
Leaf Node

Spam
Spam
Spam

↓

Pure Node ✔
```

---

### DecisionTreeRegressor

Stops splitting when:

- Stopping parameters (`max_depth`, `min_samples_split`, `min_samples_leaf`) are reached, or
- No further split significantly reduces prediction error.

Unlike classification, leaf nodes contain **continuous values**, so they are not based on class purity.

| `class_weight` | Handles imbalanced datasets (Classifier only) |

# Pipeline in Machine Learning

A **Pipeline** is a Machine Learning utility that **combines multiple steps** (such as preprocessing and model training) into a single workflow.

It executes each step **sequentially (one after another)**, making the code cleaner, reusable, and preventing data leakage.

---

# Why Use Pipeline?

- Combines preprocessing and model into a single object.
- Executes every step in the correct order.
- Reduces repetitive code.
- Prevents data leakage.
- Makes training and prediction easier.
- Useful for GridSearchCV and Cross Validation.

---

# Working of Pipeline

```text
          X_train
             │
             ▼
   Preprocessing Step
(StandardScaler, Encoder, etc.)
             │
             ▼
   Transformed Features
             │
             ▼
      Machine Learning Model
             │
             ▼
      Prediction / Output
```

Pipeline executes **one step after another**.

---

# Syntax

```python
from sklearn.pipeline import Pipeline

pipe = Pipeline(

    steps=[

        ('scaling', StandardScaler()),

        ('model', DecisionTreeRegressor())

    ]

)

pipe.fit(X_train, y_train)
```

---

# How Pipeline Works

```text
pipe.fit(X_train, y_train)

        │

        ▼

StandardScaler.fit_transform(X_train)

        │

        ▼

Updated X_train

        │

        ▼

Step 2:
DecisionTreeRegressor.fit(updated_X_train, y_train)

        │

        ▼

Model Trained
```

---

# Example 1

Using **one preprocessing technique** followed by a model.


# Limitation of Pipeline

Pipeline executes steps **sequentially**.

It cannot apply **multiple preprocessing techniques independently** on different columns.

Example:

- Numerical Columns → Scaling

For this, we use **ColumnTransformer**.

# Using Multiple Preprocessing Techniques on the Same Columns

Sometimes we need to perform **more than one preprocessing step** on the same columns.

Example:

- Standard Scaling
- Polynomial Feature Generation

Since these steps depend on each other, they must execute **sequentially**.

# Ensemble Learning

**Ensemble Learning** is a Machine Learning technique that combines **multiple models** (called **base learners**) to improve prediction accuracy and overall model performance.

The two main types of Ensemble Learning are:

1. **Bagging (Bootstrap Aggregating)**
2. **Boosting**

> **Why use Ensemble Learning?**
>
> When a single Decision Tree does not perform well, ensemble methods can improve its performance.

---

# Bagging (Bootstrap Aggregating)

**Bagging** is an ensemble technique that trains **multiple Decision Trees independently and in parallel**, then combines their predictions.

- Trees are trained **independently**.
- Uses **Bootstrap Sampling** to create different training datasets.
- Reduces **Variance** (helps reduce overfitting).
- Does **not reduce Bias**.

> **Example:** Random Forest

---

# Random Forest

A **Random Forest** is a **Supervised Machine Learning algorithm** based on the **Bagging Ensemble** technique.

It can be used for both:

- **Classification**
- **Regression**

---

# How Random Forest Works

### Step 1: Create Multiple Decision Trees

By default, Random Forest builds **100 Decision Trees** (`n_estimators=100`).

```
Decision Tree 1

Decision Tree 2

Decision Tree 3

...

Decision Tree 100
```

---

### Step 2: Bootstrap Sampling

Each tree is trained on a **random sample of the training data** (sampling with replacement).

As a result:

- Every tree receives a slightly different dataset.
- Each tree learns different patterns.
- Trees become less correlated.

```text
Original Dataset

        │

        ▼

Bootstrap Sampling

 ┌──────┬──────┬──────┐

 ▼      ▼      ▼

Tree1  Tree2  Tree3 ... Tree100
```

---

### Step 3: Train the Trees

- All Decision Trees are trained **in parallel**.
- Every tree is **independent** of the others.

```text
             Dataset

                │

                ▼

      Bootstrap Sampling

                │

     ┌──────────┼──────────┐

     ▼          ▼          ▼

   Tree 1     Tree 2     Tree 3 ... Tree100

     │          │          │

     └──────────┼──────────┘

                ▼

      Combine Predictions
```

---

# How Random Forest Makes Predictions

## 1. Classification

Each Decision Tree predicts a class label.

The final prediction is based on **Majority Voting**.

### Example

```
Tree Predictions

Spam
Spam
Not Spam
Spam
Spam
```

Prediction Count

```
Spam = 4

Not Spam = 1
```

Final Prediction

```
Spam
```

---

## 2. Regression

Each Decision Tree predicts a continuous value.

The final prediction is the **Average (Mean)** of all tree outputs.

### Example

```
Tree Predictions

72

75

74

70

79
```

Final Prediction

```
Average

=

(72 + 75 + 74 + 70 + 79)

--------------------------

5

=

74
```

---

# When to Use Random Forest?

Use Random Forest when:

- A Decision Tree is **overfitting** the training data.
- Higher prediction accuracy is required.
- You want a more stable and robust model.

---

# When Not to Use Random Forest?

Avoid Random Forest when:

- The Decision Tree is **underfitting** (adding more trees will not solve underfitting).
- The dataset is extremely large and training time is a concern.
- A simple Decision Tree already performs well and model interpretability is important.

---

# Advantages

- Reduces overfitting.
- Reduces variance.
- More accurate than a single Decision Tree.
- Works for both classification and regression.
- Handles noisy data well.

---

# Limitations

- Higher computational cost.
- Slower training and prediction than a single Decision Tree.
- Less interpretable than a Decision Tree.
- May not be suitable for very large datasets when speed is critical.

````python
# ============================================
# Import Libraries
# ============================================

import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import classification_report


# ============================================
# Load Dataset
# ============================================

df = pd.read_csv('datasets/loan_data.csv')

print(df.head())


# ============================================
# Separate Features and Target
# ============================================

X = df.drop(columns='loan_status')

y = df['loan_status']


# ============================================
# Train Test Split
# ============================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,
    train_size=0.8,
    random_state=42

)


# ============================================
# Column Transformer
# ============================================

preprocessing = ColumnTransformer(

    transformers=[

        (
            'onehot',

            OneHotEncoder(
                sparse_output=False,
                handle_unknown='ignore'
            ),

            ['c1', 'c2', 'c3', 'c4']

        ),

        (
            'ordinal',

            OrdinalEncoder(

                categories=[
                    ['Low', 'Medium', 'High']
                ]

            ),

            ['c5']

        )

    ],

    remainder='passthrough'

)


# ============================================
# Pipeline
# ============================================

pipeline = Pipeline(

    steps=[

        ('preprocessing', preprocessing),

        ('model', RandomForestClassifier(

            n_estimators=100,

            random_state=42

        ))

    ]

)
````

````python


# ============================================
# Train Model
# ============================================

pipeline.fit(X_train, y_train)


# ============================================
# Prediction
# ============================================

train_pred = pipeline.predict(X_train)

test_pred = pipeline.predict(X_test)


# ============================================
# Classification Report
# ============================================

print("========== TRAIN REPORT ==========\n")

print(classification_report(y_train, train_pred))


print("========== TEST REPORT ==========\n")

print(classification_report(y_test, test_pred))
````

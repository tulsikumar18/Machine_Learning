# K-Means Clustering

## What is K-Means Clustering?

**K-Means** is an **Unsupervised Machine Learning Algorithm** used for **Clustering**. It groups similar data points into **K clusters** based on their similarity.

- **K** → Number of Clusters
- **Means** → Average (Centroid of a cluster)

Unlike supervised learning, K-Means does **not require a target column**.

```text
Dataset
    │
    ▼
K-Means Algorithm
    │
    ▼
Cluster 1   Cluster 2   Cluster 3
```

---

# Why do we use K-Means?

K-Means helps in grouping similar observations together.

### Real-Life Applications

- Customer Segmentation
- Market Basket Analysis
- Student Performance Grouping
- Image Compression
- News Article Clustering
- Recommendation Systems

---

# How K-Means Works?

K-Means follows the following steps.

```text
Dataset
    │
    ▼
Choose K (No. of Clusters)
    │
    ▼
Initialize Random Centroids
    │
    ▼
Calculate Distance
(Euclidean Distance)
    │
    ▼
Assign Points to Nearest Centroid
    │
    ▼
Calculate New Centroids
    │
    ▼
Repeat Until Centroids Stop Changing
```

# Step 1 : Calculate Euclidean Distance

K-Means uses the **Euclidean Distance Formula**.

### Formula

```text
Distance = √[(x₂ - x₁)² + (y₂ - y₁)²]
```

The point is assigned to the centroid with the **minimum distance**.

---

### Example

For the first row

```text
(21,7000)
```

Distance from C1

```text
√[(21-25)² + (7000-26000)²]

≈ 19000
```

Distance from C2

```text
√[(21-35)² + (7000-50000)²]

≈ 43000
```

Since

```text
19000 < 43000
```

The point belongs to

```text
Cluster 1
```

---

# Step 2 : Assign All Points

Calculate the distance of every point from every centroid.

Assign each point to the nearest centroid.

```text
Point
   │
   ▼
Nearest Centroid?
   │
   ├────────► Cluster 1
   │
   └────────► Cluster 2
```

---

# Step 3 : Update the Centroids

After assigning all points, compute the **new centroid** by taking the average of all points in each cluster.

```text
Cluster

(21,7000)

(25,26000)

(30,25000)

↓

New Centroid

(Average Age,
 Average Salary)
```

---

# Step 4 : Repeat

Again calculate distances using the **new centroids**.

Continue until the centroids no longer change.

```text
Assign Points
      │
      ▼
Calculate New Centroids
      │
      ▼
Centroids Changed?
      │
  Yes ─────► Repeat
      │
      ▼
      No
      │
      ▼
Final Clusters
```

---

# Choosing the Best Value of K

The value of **K** is selected using the **Elbow Method**.

---

# Elbow Method

The Elbow Method helps find the optimal number of clusters.

It uses **WCSS (Within Cluster Sum of Squares)**.

### Steps

1. Train K-Means for different values of K.
2. Calculate WCSS for each K.
3. Plot K vs WCSS.
4. Choose the point where the graph bends like an elbow.

```text
WCSS
 ^
 |
 |\
 | \
 |  \
 |   \__
 |      \____
 |
 +------------------------>
     1 2 3 4 5 6
          ↑
      Best K
```

# Syntax

```python
from sklearn.cluster import KMeans

cluster = KMeans(
    n_clusters=2,
    random_state=42
)

labels = cluster.fit_predict(df)
```

### Explanation

- `n_clusters` → Number of clusters.
- `fit()` → Learns the cluster centroids.
- `predict()` → Assigns cluster labels to new data.
- `fit_predict()` → Performs both operations together.

---

# Why Scaling is Important?

K-Means uses **Euclidean Distance**.

If features have different scales, larger values dominate the distance calculation.

Example

```text
Age

20

25

30

Salary

20000

50000

100000
```

Salary has much larger values than Age, so it will dominate the distance.

Therefore, **feature scaling (StandardScaler or MinMaxScaler)** is recommended before applying K-Means.

---

# Can K-Means Handle Categorical Data?

No.

K-Means is designed for **numerical features** because it relies on Euclidean Distance.

If your dataset contains categorical columns:

- Apply **Encoding** (One-Hot Encoding, Ordinal Encoding, etc.).
- Scale numerical features if necessary.

---

# Advantages

- Simple and easy to understand.
- Fast for large datasets.
- Efficient clustering algorithm.
- Easy to implement.
- Works well for well-separated clusters.

---

# Disadvantages

- Value of **K** must be chosen manually.
- Sensitive to outliers.
- Sensitive to initial centroid selection.
- Works only with numerical data.
- Assumes clusters are roughly spherical.

# Key Points to Remember

- Unsupervised Learning Algorithm.

````python
## download one dataset, and  use columnTransformer and pipeline..to implement KMeans..
````

````python
import pandas as pd

from sklearn.cluster import KMeans
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.impute import SimpleImputer
````

````python
df = pd.read_csv('/Users/tulsikumaryadav/Desktop/Machine_learning/datasets/Mall_Customers.csv')
df.head()
````

````python
df.info()
````

````python
X = df.drop(columns='CustomerID')
````

````python
num_cols = X.select_dtypes(include = 'number').columns

cat_cols = X.select_dtypes(include='object').columns

````

````python
## pipeline


num_pipeline = Pipeline(

    steps=[
        ('imputer', SimpleImputer(strategy = 'mean')),
        ('scaler', StandardScaler())
    ]
)

cat_pipeline = Pipeline(

    steps = [

        ('imputer', SimpleImputer(strategy = 'most_frequent') ),
        ('encoder', OneHotEncoder(handle_unknown = 'ignore'))
    ]
)
````

````python
preprocessing = ColumnTransformer(

    transformers=[

        ("num", num_pipeline, num_cols),

        ("cat", cat_pipeline, cat_cols)

    ]

)
````

````python
main_pipeline = Pipeline(

    steps=[

        ("preprocessing", preprocessing),

        ("model", KMeans(
            n_clusters=2,
            random_state=42    # helps in selecting the same centroid..
        ))

    ]

)
````

````python
main_pipeline.fit(df)
````

````python
# Predict Clusters
clusters = main_pipeline.predict(X)

# Add Cluster Column
df["Cluster"] = clusters

print(df) # shows , in which clusters , it will come..
````

````python
## and by using fit_predict() , gives me the clusters values.
main_pipeline.fit_predict(X)
````

````python
## by using fit_transform() we can get the cluster distance.. for each row..


main_pipeline.fit_transform(X)
````

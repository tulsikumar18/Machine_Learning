## WHat is resampling technique?
It is a preprocessing technique it balances the data by adding or removing samples and we use this only for train data.

## Types:
- Undersampling
- Oversampling

## Under Sampling
- It is a resampling preprocessing technique which removes the samples randomly from majority class labels to balance the dataset.

> Example 1:
- 0 -> 700  |  0 -> 300
- 1 -> 300  |  1 -> 300

> Example 2: 
- 0 -> 1000
- 1 -> 9000   we should not perform undersampling for this type of data

> Example 3:
- 0 -> 100000
- 1 -> 100

- In real time we use both undersampling and oversampling -> parameter 

## parameter
1. sampling_strategy = 0.2  (use 0.2 - 0.5)
- sampling strategy we use when we have huge values.
- if we increase sampling_strategy, then it will decrease then majority class labels.
- if we use value > 0.5 then number of samples will decrease and we will lose important data.

### Formula: 

$$  minority samples/majority new samples = 0.2  $$

## Oversampling
It is a resampling preprocessing technique which adds the samples randomly to minority class labels to balance the dataset.
- it uses SMOTE (Synthetic Minority Oversampling TEchnique)
- Internally it uses KNN(distance formula)

Example 1:
- 0 -> 700 -> 700
- 1 -> 300 -> 700



---

when a class has 100000 samples and other has 100 samples then if we use oversampling, model will become overfit.

## SOLUTION:

- Use undersampling and oversampling one after the other.
- first undersampling and then oversampling
>100000 -> 20000 -> 20000

>100 -> 100 -> 20000

# Very Very Very Important 
## Why we use SMOTE
- IT is a oversampling preprocessing technique,how it adds samples using KNN

## WHen we use SMOTE?
When the data is very important, do not use undersampling directly, use SMOTE

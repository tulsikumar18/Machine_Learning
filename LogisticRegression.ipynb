# Logistic Regression - Sigmoid Function

## What is Logistic Regression?

- Logistic Regression is a **Supervised Machine Learning Algorithm** used for **Binary Classification** problems.
- It predicts the probability that an input belongs to one of two classes (**0 or 1**).
- Although its name contains **Regression**, it is mainly used for **Classification**.
- It is called **Regression** because it first computes a linear equation and then converts it into a probability using the **Sigmoid Function**.

---

# Linear Equation

Logistic Regression first computes a linear combination of all input features.

### Formula

```text
z = w1x1 + w2x2 + w3x3 + ... + wnxn + b
```

Where:

- **x1, x2, ..., xn** → Input Features
- **w1, w2, ..., wn** → Weights (Importance of Features)
- **b** → Bias (Intercept)
- **z** → Linear Output

---

# Sigmoid Function

The linear output (**z**) is passed to the **Sigmoid Function**.

### Formula

```text
Sigmoid(z) = 1 / (1 + e^(-z))
```

where

```text
z = w1x1 + w2x2 + ... + wnxn + b
```

---

## Why Sigmoid Function?

The sigmoid function converts any real number into a probability between **0 and 1**.

- Large Positive z → Probability approaches **1**
- Large Negative z → Probability approaches **0**
- z = 0 → Probability = **0.5**

# Key Points

- Logistic Regression is used for **Binary Classification**.
- It predicts **probabilities**, not continuous values.
- The **Sigmoid Function** converts any real number into a value between **0 and 1**.
- The sigmoid curve is **S-shaped**.
- The default classification threshold is **0.5**.
- The weights (**w**) represent the importance of each feature.
- The bias (**b**) shifts the decision boundary.
- Logistic Regression internally computes a **linear equation**, which is why it is called "Regression".

---

# Why Not Use Linear Regression for Classification?

Although Linear Regression is excellent for predicting **continuous values**, it is **not suitable** for classification.

---

## 1. Output Can Be Outside the Range (0 to 1)

Linear Regression predicts values from

```text
-∞ to +∞
```

Example

```text
Prediction = -2.4

Prediction = 3.8
```

These are **not valid probabilities**.

---

## 2. No Probability Interpretation

Linear Regression predicts a continuous number.

Classification requires the **probability** that an observation belongs to a class.

Example

```text
Linear Regression

Prediction = 2.3
```

```text
Logistic Regression

Probability = 0.87
```

---

## 3. No Proper Decision Boundary

Classification needs a threshold (usually **0.5**).

Linear Regression has **no built-in probability threshold**, making classification unreliable.

---

## 4. Sensitive to Outliers

Linear Regression minimizes the **Mean Squared Error (MSE)**.

A few outliers can significantly change the regression line, reducing classification performance.

---

## 5. Poor Fit for Binary Classes

Binary classification contains only two labels.

```text
0 → Negative

1 → Positive
```

A straight regression line cannot properly model these classes.

Logistic Regression uses an **S-shaped curve**, which better separates the two classes.

---

## 6. Logistic Regression Produces Valid Probabilities

Logistic Regression applies the Sigmoid Function.

```text
Sigmoid(z) = 1 / (1 + e^(-z))
```

Therefore,

```text
Any Input

↓

Probability (0 to 1)
```

making it ideal for binary classification.

# Regularization

Regularization is a **Machine Learning technique** used to **reduce overfitting (high variance)** by adding a **penalty** to the model's loss function. It prevents the model from learning unnecessary patterns (noise) from the training data, resulting in better performance on unseen data.

---

## Why do we need Regularization?

Sometimes a model performs very well on the training data but poorly on new (unseen) data. This is known as **Overfitting**.

Regularization helps by:

- Reducing model complexity.
- Preventing overfitting.
- Improving generalization.
- Controlling the feature weights (coefficients).

---

## Generalization

**Generalization** refers to the ability of a machine learning model to perform well on **unseen data**.

- Good Generalization → Performs well on both training and testing data.
- Poor Generalization → Overfits the training data and performs poorly on testing data.

---

# Types of Regularization

There are three commonly used regularization techniques:

- **L1 Regularization (Lasso)**
- **L2 Regularization (Ridge)**
- **Elastic Net (L1 + L2)**

---

# 1. Lasso Regression (L1 Regularization)

Lasso Regression adds an **L1 penalty** to the loss function.

Its main purpose is **Feature Selection**.

It pushes some feature weights **exactly to zero**, effectively removing those features from the model.

### Example

Regression Equation

```text
y = m₁x₁ + m₂x₂ + c
```

Suppose

```text
m₁ = 0
m₂ = 5
```

Prediction becomes

```text
y = (0 × x₁) + (5 × x₂) + c
```

Since **m₁ = 0**, feature **x₁** has no contribution and is automatically removed from the model.

---

## When to use Lasso?

- Dataset has many features.
- Feature selection is required.
- Some features may be irrelevant.
- High-dimensional datasets (e.g., NLP/Text Classification).

---

## Advantages

- Performs automatic feature selection.
- Reduces overfitting.
- Produces simpler and more interpretable models.

---

## Limitation

- Important correlated features may also be removed.

---

# 2. Ridge Regression (L2 Regularization)

Ridge Regression adds an **L2 penalty** to the loss function.

Instead of removing features, it **shrinks their weights**.

> **Important:** Ridge reduces the weights but **never makes them exactly zero**.

### Example

```text
Before

m₁ = 8

m₂ = 6

m₃ = 9
```

After Ridge

```text
m₁ = 3.5

m₂ = 2.8

m₃ = 4.1
```

All features remain in the model, but their influence is reduced.

---

## When to use Ridge?

- All features are important.
- Multicollinearity exists between features.
- Feature selection is not required.

---

## Advantages

- Handles multicollinearity effectively.
- Keeps all features.
- Reduces overfitting.

---

## Limitation

- Does not perform feature selection.

---

# 3. Elastic Net

Elastic Net combines both **L1 (Lasso)** and **L2 (Ridge)** regularization.

It performs **feature selection** while also handling **multicollinearity**.

---

## L1 Ratio

Elastic Net uses the parameter **l1_ratio**.

| l1_ratio | Meaning |
|----------|---------|
| 0 | 100% Ridge (L2) |
| 1 | 100% Lasso (L1) |
| 0.3 | 30% L1 + 70% L2 |
| 0.5 | 50% L1 + 50% L2 |
| 0.8 | 80% L1 + 20% L2 |

---

## When to use Elastic Net?

- Large datasets with many features.
- Some features are important, while others are not.
- Dataset has multicollinearity.
- Need both feature selection and weight shrinking.

---

## Advantages

- Combines benefits of Lasso and Ridge.
- Performs feature selection.
- Handles multicollinearity.
- Reduces overfitting.

---

# Comparison of Regularization Techniques

| Feature | Lasso (L1) | Ridge (L2) | Elastic Net |
|----------|------------|------------|-------------|
| Feature Selection | ✅ Yes | ❌ No | ✅ Yes |
| Weight becomes Zero | ✅ Yes | ❌ No | Sometimes |
| Handles Multicollinearity | ❌ Limited | ✅ Yes | ✅ Yes |
| Keeps All Features | ❌ No | ✅ Yes | Partially |
| Best For | Many Features | All Features Important | Mixed Scenarios |

# Gradient Descent

Gradient Descent is an **Optimization Algorithm** used to find the **best weights (coefficients)** that minimize the loss (error) of a machine learning model.

The objective is to reduce the model's prediction error by updating the weights iteratively.

---

## Why do we need Gradient Descent?

Instead of manually finding the best weights, Gradient Descent automatically adjusts them until the model reaches the minimum loss.

---

## How Gradient Descent Works

```text
Randomly Initialize Weights
            │
            ▼
Make Predictions
            │
            ▼
Calculate Loss
            │
            ▼
Compute Gradient
            │
            ▼
Update Weights
            │
            ▼
Repeat Until Loss is Minimum
```

---

## Where is Gradient Descent used?

- Linear Regression (when solved iteratively)
- Logistic Regression
- Neural Networks
- XGBoost (gradient boosting framework)
- Lasso Regression
- Ridge Regression
- Elastic Net

# Loss Function

A **Loss Function** measures how much error the model makes for **a single training example**.

Lower loss indicates better predictions.

In Logistic Regression, the commonly used loss function is **Log Loss (Binary Cross-Entropy Loss)**.

---

## Log Loss Formula

```text
Loss = -( y log(p) + (1-y) log(1-p) )
```

Where:

- **y** = Actual Label (0 or 1)
- **p** = Predicted Probability

---

# Cost Function

The **Cost Function** is the **average loss** calculated over the **entire training dataset**.

```text
Cost = Average of all Loss Values
```

Gradient Descent minimizes the **Cost Function**, not just the loss of one sample.

---

# Difference between Loss Function and Cost Function

| Loss Function | Cost Function |
|---------------|---------------|
| Error for one training example | Average error for the entire dataset |
| Used for a single prediction | Used to optimize the complete model |
| Smaller value is better | Smaller value is better |

---

# Logistic Regression Parameters

Logistic Regression is a **Supervised Machine Learning Classification Algorithm** used to classify data into **two or more classes**.

It internally uses **Gradient Descent** to optimize the feature weights and minimize the **Log Loss (Cost Function)**.

---

# Creating the Model

```python
from sklearn.linear_model import LogisticRegression

```

# Important Parameters

## 1. penalty

The **penalty** parameter specifies the type of **Regularization** used to reduce overfitting.

By default, Logistic Regression uses **L2 (Ridge) Regularization**.

### Syntax

```python
model = LogisticRegression(
    penalty="l2"
)
```

### Available Options

| Penalty | Regularization | Purpose |
|----------|---------------|---------|
| `"l1"` | Lasso | Performs feature selection by making some weights exactly zero |
| `"l2"` | Ridge (Default) | Shrinks weights and reduces overfitting |
| `"elasticnet"` | L1 + L2 | Combines feature selection and weight shrinking |
| `None` | No Regularization | Model is trained without any penalty |

---

### Example

```python
model = LogisticRegression(
    penalty="l1",
    solver="liblinear"
)
```

> **Interview Point:** Every penalty is **not supported by every solver**.

---

# 2. solver

A **Solver** is an optimization algorithm that helps **Gradient Descent** find the best weights by minimizing the **Log Loss**.

Default Solver

```python
solver="lbfgs"
```

---

## Supported Solvers

| Solver | Supported Penalty | Multiclass Support |
|---------|-------------------|--------------------|
| `lbfgs` | L2, None | ✅ Yes |
| `liblinear` | L1, L2 | ❌ No (Binary only) |
| `newton-cg` | L2, None | ✅ Yes |
| `newton-cholesky` | L2, None | ❌ No |
| `sag` | L2, None | ✅ Yes |
| `saga` | L1, L2, ElasticNet, None | ✅ Yes |

---

## Multiclass Classification

For **Binary Classification**, Logistic Regression uses the **Sigmoid Function**.

For **Multiclass Classification**, supported solvers internally use the **Softmax Function**.

```text
Binary Classification

Sigmoid Function

↓

Spam / Not Spam
```

```text
Multiclass Classification

Softmax Function

↓

Class 0

Class 1

Class 2

...
```

---

## Important Note

The **SAG** and **SAGA** solvers work best when all numerical features are on a **similar scale**.

Therefore, always apply **feature scaling** before using these solvers.

Example

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

---

## When to Use SAG and SAGA?

- Large datasets
- Many feature columns
- Faster convergence after scaling

---

# 3. max_iter

`max_iter` specifies the **maximum number of iterations** the solver can perform to find the optimal weights.

Default

```python
max_iter = 100
```

Example

```python
model = LogisticRegression(
    max_iter=500
)
```

If the solver cannot converge within the specified iterations, Scikit-learn displays a warning.

---

# 4. tol (Tolerance)

Tolerance determines **when the optimization process should stop**.

Default

```python
tol = 0.0001
```

If the improvement in loss becomes smaller than the tolerance value, Gradient Descent stops updating the weights.

> **Note:** The default value works well for most problems, so it is rarely changed.

---

# 5. class_weight

`class_weight` is used to handle **imbalanced datasets**.

Default

```python
class_weight=None
```

---

```

Not Spam  = 480
```

The classes are nearly equal.

---

## Imbalanced Dataset

```
Spam      = 800

Not Spam  = 200
```

The model may become biased toward the majority class.

---

## Handling Imbalanced Data

### Method 1

```python
model = LogisticRegression(

    class_weight="balanced"

)
```

The algorithm automatically assigns higher importance to the minority class.

---

### Method 2

- Oversampling
- Undersampling
- SMOTE

# 6. C (Regularization Strength)

`C` controls the **strength of regularization**.

Default

```python

```

`C` is the **inverse of regularization strength**.

### Interpretation

| C Value | Regularization | Model |
|----------|---------------|-------|
| Small C (0.01, 0.1) | Strong | Simpler model (may underfit) |
| Default C (1.0) | Moderate | Balanced model |
| Large C (10, 100) | Weak | Complex model (may overfit) |

---

# Example

```python
)
```

# 1. Confusion Matrix

A **Confusion Matrix** compares the **Actual Labels** with the **Predicted Labels**.

It tells us:

- Correct Predictions
- Incorrect Predictions
- Types of Errors

---

## Structure

> Actual Labels → **Rows**

> Predicted Labels → **Columns**

```text
                    Predicted

                 Negative   Positive

Actual Negative     TN         FP

Actual Positive     FN         TP
```

### Meaning

- **TP (True Positive)** → Actual Positive predicted as Positive
- **TN (True Negative)** → Actual Negative predicted as Negative
- **FP (False Positive)** → Actual Negative predicted as Positive (Type-I Error)
- **FN (False Negative)** → Actual Positive predicted as Negative (Type-II Error)

---

## Example

```text
                Predicted

             0          1

Actual 0     80         20

Actual 1    100        300
```

```
TN = 80
FP = 20
FN = 100
TP = 300
```

Correct Predictions

```
TN + TP = 80 + 300 = 380
```

Incorrect Predictions

```
FP + FN = 20 + 100 = 120
```

> **Good Model:** Higher values should lie on the **diagonal (TN & TP)**.

---

## Python Code

```python
from sklearn.metrics import confusion_matrix

y_pred = model.predict(X_test)

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

---

# 2. Precision Score

Precision tells:

> **Out of all the samples predicted as Positive, how many are actually Positive?**

It measures the **quality of positive predictions**.

---

## Formula

```text
Precision = TP / (TP + FP)
```

---

## Example

```
TP = 300
FP = 20
```

```
Precision

= 300 / (300 + 20)

= 300 / 320

= 93.75%
```

### Conclusion

Out of all the **predicted Positive samples**, **93.75% are actually Positive**, while **6.25% are False Positives**.

---

## Precision for Negative Class

When `pos_label = 0`

```text
Precision = TN / (TN + FN)
```

---

## Python Code

```python
from sklearn.metrics import precision_score

# Positive Class (Default)
precision_score(y_test, y_pred)

# Negative Class
precision_score(y_test, y_pred, pos_label=0)
```

---

## Use Precision When

False Positives are costly.

Examples:

- Spam Detection
- Fraud Detection
- Recommendation Systems

---

# 3. Recall Score

Recall tells:

> **Out of all the Actual Positive samples, how many are correctly predicted?**

It measures the model's ability to detect positive cases.

---

## Formula

```text
Recall = TP / (TP + FN)
```

---

## Example

```
TP = 300
FN = 100
```

```
Recall

= 300 / (300 + 100)

= 75%
```

### Conclusion

Out of all the **Actual Positive samples**, **75% are correctly predicted**, while **25% are missed (False Negatives)**.

---

## Recall for Negative Class

When `pos_label = 0`

```text
Recall = TN / (TN + FP)
```

---

## Python Code

```python
from sklearn.metrics import recall_score

# Positive Class
recall_score(y_test, y_pred)

# Negative Class
recall_score(y_test, y_pred, pos_label=0)
```

---

## Use Recall When

False Negatives are costly.

Examples:

- Cancer Detection
- Disease Prediction
- Fraud Detection
- Fire Alarm Systems

---

# Precision vs Recall

| Precision | Recall |
|------------|---------|
| Out of predicted positives, how many are correct? | Out of actual positives, how many are correctly predicted? |
| Focuses on False Positives | Focuses on False Negatives |
| Prediction Quality | Detection Ability |

---

# 4. Accuracy Score

Accuracy tells:

> **Out of all samples, how many are correctly classified?**

---

## Formula

```text
Accuracy

=

(TP + TN)

/

(TP + TN + FP + FN)
```

---

## Example

```text
                Predicted

             0          1

Actual 0    900        100

Actual 1     50        950
```

```
Accuracy

=

(900 + 950)

/

2000

=

92.5%
```

---

## Why Accuracy is Misleading for Imbalanced Data?

Dataset

```
Negative = 1000

Positive = 200
```

Prediction

```text
                Predicted

             0          1

Actual 0    1000        0

Actual 1     200        0
```

```
Accuracy

=

1000 / 1200

=

83.3%
```

But

```
Recall (Positive)

=

0 / (0 + 200)

=

0%
```

Although Accuracy is **83%**, the model **fails to detect every positive sample**.

> **Conclusion:** Accuracy is not reliable for **Imbalanced Datasets**.

---

## Python Code

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

---

## Use Accuracy When

- Dataset is Balanced
- Both classes have nearly equal samples

---

# 5. F1 Score

F1 Score combines **Precision** and **Recall** into a single metric.

It is the **Harmonic Mean** of Precision and Recall.

---

## Formula

```text
F1 Score

=

2 × Precision × Recall

-----------------------

Precision + Recall
```

---

## Example

```
Precision = 0.80

Recall = 0.90
```

```
F1

=

2 × 0.80 × 0.90

----------------

0.80 + 0.90

=

0.847

=

84.7%
```

---

## Imbalanced Dataset Example

```text
                Predicted

             0          1

Actual 0    1000        0

Actual 1     200        0
```

```
Precision = 0

Recall = 0

F1 Score = 0
```

Although **Accuracy = 83%**, **F1 Score = 0**, indicating the model performs poorly.

---

## Python Code

```python
from sklearn.metrics import f1_score

# Positive Class
f1_score(y_test, y_pred)

# Negative Class
f1_score(y_test, y_pred, pos_label=0)
```

---

## Use F1 Score When

- Dataset is Imbalanced
- Both Precision and Recall are equally important

Examples:

- Spam Detection
- Medical Diagnosis
- Fraud Detection
- Intrusion Detection

---

# Accuracy vs F1 Score

| Accuracy | F1 Score |
|-----------|----------|
| Best for Balanced Dataset | Best for Imbalanced Dataset |
| Measures Overall Correctness | Balances Precision & Recall |
| Can be Misleading | More Reliable |

---

# Choosing the Right Metric

```text
Balanced Dataset
        │
        ▼
Use Accuracy Score

------------------------

Imbalanced Dataset
        │
        ▼
Use F1 Score
```

---

# Summary Table

| Metric | Formula | Best Used When |
|---------|---------|----------------|
| Confusion Matrix | TN, FP, FN, TP | Base for all metrics |
| Precision | TP / (TP + FP) | False Positives are costly |
| Recall | TP / (TP + FN) | False Negatives are costly |
| Accuracy | (TP + TN) / Total | Balanced Dataset |
| F1 Score | 2 × (Precision × Recall) / (Precision + Recall) | Imbalanced Dataset |

---

**Answer:** Rows

Because it may produce a high score even if the model completely ignores the minority class.

---

### Q6. When should F1 Score be used?

When the dataset is **Imbalanced** and both Precision and Recall are important.

---

# Best Practices

- Start evaluation with the **Confusion Matrix**.
- Use **Precision** when minimizing False Positives is important.
- Use **Recall** when minimizing False Negatives is important.
- Use **Accuracy** only for Balanced datasets.
- Use **F1 Score** for Imbalanced datasets.
- Compare multiple metrics instead of relying on a single metric.

# Column Transformer (Feature Preprocessing Technique)

## What is Column Transformer?

**Column Transformer** is a **Machine Learning preprocessing technique** used to apply **multiple preprocessing techniques** to different feature columns **at the same time**.

It allows us to preprocess different columns in **parallel** and combines the transformed output into a single dataset.

> **Note:** Column Transformer is used **only for preprocessing techniques**, **not for Machine Learning algorithms**.

---

# Why do we need Column Transformer?

In a real-world dataset, different columns often require different preprocessing techniques.

Example:

| Feature | Preprocessing Required |
|---------|------------------------|
| Age | MinMaxScaler |
| Salary | RobustScaler |
| City | OneHotEncoder |
| Occupation | TargetEncoder |

Without a Column Transformer, we would need to preprocess each column separately.

With a Column Transformer, all preprocessing steps are applied together in a single pipeline.

---

# Workflow

```text
              Input Features
                    │
      ┌─────────────┼─────────────┐
      │             │             │
      ▼             ▼             ▼
 MinMaxScaler  RobustScaler  TargetEncoder
      │             │             │
      └─────────────┼─────────────┘
                    │
                    ▼
         Column Transformer
                    │
                    ▼
     Combined Transformed Dataset
```

---

# When to Use Column Transformer?

Use a Column Transformer when different feature columns require different preprocessing techniques.

Example:

- Numerical columns → Scaling
- Categorical columns → Encoding
- Text columns → Vectorization (if required)

---

# Common Preprocessing Techniques

| Technique | Best Used For |
|-----------|---------------|
| MinMaxScaler | Numerical data with **few or no outliers** |
| RobustScaler | Numerical data with **many outliers** |
| StandardScaler | Normally distributed numerical data |
| OneHotEncoder | Nominal categorical features |
| OrdinalEncoder | Ordered categorical features |
| TargetEncoder | High-cardinality categorical features |

---

# Syntax

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessing = ColumnTransformer(

    transformers=[

        ('name1', preprocessing_object, column_names),

        ('name2', preprocessing_object, column_names)

    ],

    remainder='drop'   # or 'passthrough'

)
```

---

# Example

Suppose the dataset contains:

| Feature | Preprocessing |
|---------|---------------|
| f1, f2 | MinMaxScaler |
| f3, f4 | RobustScaler |
| f5, f6 | TargetEncoder |

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, RobustScaler
from category_encoders import TargetEncoder

preprocessing = ColumnTransformer(

    transformers=[

        ('minmax_scaling', MinMaxScaler(), ['f1', 'f2']),

        ('robust_scaling', RobustScaler(), ['f3', 'f4']),

        ('target_encoding', TargetEncoder(), ['f5', 'f6'])

    ]

)
```

---

# Applying the Transformation

### Training Data

```python
X_train_new = preprocessing.fit_transform(X_train, y_train)
```

> **Note:** `TargetEncoder` requires the target column (`y_train`) during `fit_transform()`.

### Testing Data

```python
X_test_new = preprocessing.transform(X_test)
```

Only `transform()` is used because the preprocessing parameters have already been learned from the training data.

---

# How Column Transformer Works

Suppose:

- MinMaxScaler returns **2 columns**
- RobustScaler returns **2 columns**
- TargetEncoder returns **2 columns**

The Column Transformer combines all transformed columns into a single dataset.

```text
MinMaxScaler
     │
     ▼
2 Columns

RobustScaler
     │
     ▼
2 Columns

TargetEncoder
     │
     ▼
2 Columns

        │
        ▼

Column Transformer

        │
        ▼

Final Dataset (6 Columns)
```

---

# `remainder` Parameter

The `remainder` parameter controls what happens to the columns **not mentioned** in the `transformers` list.

---

## 1. `remainder='drop'` (Default)

Columns that are **not specified** are removed.

Example

Dataset

| f1 | f2 | f3 | f4 | f5 | f6 |
|----|----|----|----|----|----|

Only these columns are transformed:

```python
['f1', 'f2', 'f3', 'f4']
```

Result

```text
Only transformed columns are returned.

f5 and f6 are dropped.
```

---

## 2. `remainder='passthrough'`

Columns that are **not specified** are kept unchanged.

```python
preprocessing = ColumnTransformer(

    transformers=[

        ('scaler', MinMaxScaler(), ['f1', 'f2'])

    ],

    remainder='passthrough'

)
```

Output

```text
Scaled Columns

+

Remaining Columns

↓

Final Dataset
```

All original columns are preserved.

---

# Advantages

- Applies multiple preprocessing techniques in one step.
- Executes preprocessing in parallel.
- Combines transformed columns into one dataset.
- Reduces repetitive preprocessing code.
- Easy to integrate with machine learning pipelines.
- Makes preprocessing more organized and maintainable.

---

# Limitations

- Used only for **feature preprocessing**.
- Cannot be used to train machine learning models.
- Requires correct column selection for each transformer.

---

# Best Practices

- Apply preprocessing **after Train-Test Split**.
- Use `fit_transform()` only on the training data.
- Use `transform()` on testing or new data.
- Choose the preprocessing technique based on the type of feature.
- Use `TargetEncoder` only when high-cardinality categorical features are present.

---

```text
        ▼
Different Preprocessing
        │
        ▼

        │
        ▼
One Combined Dataset
```

# Decision Tree

A **Decision Tree** is a **Supervised Machine Learning algorithm** used for both **Classification** and **Regression**. It makes predictions by asking a **series of decision-based questions**, where each answer splits the dataset into smaller groups until a final prediction (leaf node) is reached.

> **Idea:** The tree keeps splitting the data until the nodes become as **pure** as possible.

---

# Applications

- Email Spam Detection
- Disease Prediction
- Loan Approval
- Customer Churn Prediction
- House Price Prediction (Regression)

---

# Important Terms

### 1. Root Node
- The **first (top-most)** node of the tree.
- Represents the **best feature** selected for the first split.

---

### 2. Internal Node

- Intermediate nodes that contain decision questions.
- An internal node can further split into child nodes.

> Internal nodes are generally **Impure Nodes**.

---

### 3. Leaf Node

- Final node of the tree.
- Gives the final prediction/output.
- Cannot be split further.

> Leaf nodes are generally **Pure Nodes**.

---

### 4. Split

A **Split** is the process of dividing a node into child nodes based on a feature.

> Decision Trees generally perform **Binary Splits** (2 branches).

---

### 5. Purity

A node containing samples of **only one class**.

Example

| Email | Label |
|-------|-------|
| Spam | Spam |
| Spam | Spam |
| Spam | Spam |

This is a **Pure Node** because every sample belongs to the same class.

---

### 6. Impurity

A node containing **multiple class labels**.

Example

| Email | Label |
|-------|-------|
| Spam | Spam |
| Offer | Spam |
| Meeting | Not Spam |

This is an **Impure Node** because both classes are present.

---

# Relationship

```text
Internal Node
      │
      ▼
 Impure Node
      │
 Split Again
      │
      ▼
 Pure Node
      │
      ▼
 Leaf Node
```

---

# General Structure of a Decision Tree

```text
      │       │         │       │
   Internal  Leaf    Internal  Leaf
     Node    Node      Node    Node
      │
   ┌──┴──┐
   │     │
 Leaf   Leaf

```

```text
Complete Dataset
        │
        ▼

        │

        │

        │

   │          │

```

# How Does a Decision Tree Select the Best Feature?

Decision Trees select the **best feature** using the **Criterion** parameter.

The feature with the **lowest impurity (or highest information gain)** becomes the **Root Node** or an **Internal Node**.

---

# Criterion

Criterion is a measure used to determine **which feature creates the best split**.

There are two commonly used criteria:

### 1. Gini Impurity (Default)

- Based on **probability**.
- Faster to compute.
- Default criterion in Scikit-Learn.

```python
criterion="gini"
```

---

### 2. Entropy

- Based on **Information Theory**.
- Uses **log₂**.
- Measures the amount of uncertainty in the data.

```python
criterion="entropy"
```

---

## Comparison

| Gini | Entropy |
|------|----------|
| Uses Probability | Uses log₂ |
| Faster | Slightly Slower |
| Default in Scikit-Learn | Optional |
| Measures Impurity | Measures Information Gain |

---

# Splitter Parameter

The **Splitter** determines **how candidate split points are selected**.

There are two options:

1. **best** (Default)
2. **random**

---

## 1. Splitter = "best"

The algorithm evaluates **all possible split points** and chooses the one that gives the **best criterion value**.

Example

Feature Values

```
10
15
17
8
5
5
```

### Step 1

Find unique values

```
5
8
10
15
17
```

### Step 2

Sort

```
5
8
10
15
17
```

### Step 3

Create possible ranges

```
5 - 8

8 - 10

10 - 15

15 - 17
```

### Step 4

Find midpoint of each range

```
6.5

9

12.5

16
```

### Step 5

For every midpoint,

the Decision Tree calculates the **Criterion (Gini/Entropy)**.

The split having the **lowest impurity** (or highest information gain) is selected as the **Best Split**.

---

## Workflow

```text
Feature Values
      │
      ▼
Unique Values
      │
      ▼
Sort Values
      │
      ▼
Generate Midpoints
      │
      ▼
Calculate Gini / Entropy
      │
      ▼
Choose Best Split
```

---

## 2. Splitter = "random"

Instead of checking all possible split points,

the algorithm randomly selects one of the possible split points.

This is generally faster but may not always produce the optimal tree.

```python
splitter="random"
```

---

# Best vs Random Splitter

| Best | Random |
|------|---------|
| Evaluates all split points | Chooses a random split |
| Better accuracy | Faster training |
| More computation | Less computation |
| Default option | Used for randomness |

---

# Python Example

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(

    criterion="gini",      # or "entropy"

    splitter="best",       # or "random"

    random_state=42

)

model.fit(X_train, y_train)
```

---

# Key Points

- Decision Tree is a **Supervised Learning** algorithm.
- Used for both **Classification** and **Regression**.
- Makes predictions by asking a series of questions.
- The first selected feature becomes the **Root Node**.
- Internal nodes are generally **Impure Nodes**.
- Leaf nodes are generally **Pure Nodes**.
- Splitting continues until nodes become pure or stopping conditions are met.
- **Criterion** selects the best feature for splitting.
- **Gini** uses probability.
- **Entropy** uses log₂ and information gain.
- **Splitter = best** checks all possible split points.
- **Splitter = random** chooses a random split point.

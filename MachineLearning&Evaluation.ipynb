# Machine Learning (ML)

Machine Learning (ML) is a **subset of Artificial Intelligence (AI)** that enables computers to **learn patterns from historical data** and make predictions or decisions **without being explicitly programmed**.

### Flow of Machine Learning

```text
Historical Data
        │
        ▼
Learn Patterns
        │
        ▼
Train Model
        │
        ▼
Prediction on New Data
```

---

# Types of Machine Learning

There are four main types of Machine Learning:

1. Supervised Learning
2. Unsupervised Learning
3. Semi-Supervised Learning
4. Reinforcement Learning

---

# 1. Supervised Learning

Supervised Learning is a Machine Learning technique that learns from **labelled data**, where both the **input (X)** and **output (y)** are available.

```text
Input (X) + Output (y)
          │
          ▼
 Train the Model
          │
          ▼
 Predict Output
```

### Types of Supervised Learning

### A. Regression

Used when the **target/output is continuous (numerical)**.

Examples:

- House Price Prediction
- Salary Prediction
- Study Hours vs Marks
- Temperature Prediction

---

### B. Classification

Used when the **target/output is categorical**.

Examples:

- Email Spam Detection
- Disease Prediction
- Gender Classification
- Loan Approval

---

# 2. Unsupervised Learning

Unsupervised Learning learns patterns from **unlabelled data**, where only the **input (X)** is available.

```text
Input (X)
    │
    ▼
Find Hidden Patterns
    │
    ▼
Clusters / Groups
```

### Common Task

- Clustering

Examples:

- Customer Segmentation
- Product Recommendation
- Similar Movie Recommendation
- Market Basket Analysis

---

# 3. Semi-Supervised Learning

Semi-Supervised Learning uses **a small amount of labelled data** and **a large amount of unlabelled data**.

Example

```
Labelled Data    : 3,000

Unlabelled Data : 700,000
```

Instead of manually labelling all the data, the model learns from the labelled samples and predicts labels for the remaining data.

```text
Few Labelled Data
        +
Many Unlabelled Data
        │
        ▼
Learn Patterns
        │
        ▼
Label Remaining Data
        │
        ▼
Train Final Model
```

Examples

- Image Classification
- Medical Image Analysis
- Speech Recognition

---

# 4. Reinforcement Learning

Reinforcement Learning is a Machine Learning technique where an **Agent learns by interacting with an environment** and receives **Rewards or Penalties** for its actions.

```text
Environment
      │
      ▼
Agent Takes Action
      │
      ▼
Reward / Penalty
      │
      ▼
Learns Better Action
```

Examples

- Self-driving Cars
- Robot Navigation
- Chess Playing AI
- Game Playing (Atari, AlphaGo)

> **Note:** ChatGPT itself is **not** a Reinforcement Learning algorithm, but modern language models are often **fine-tuned using Reinforcement Learning from Human Feedback (RLHF).**

# Correlation

Correlation measures the **strength and direction of the relationship** between two numerical variables.

The correlation coefficient ranges between

```text
-1  to  +1
```

- **+1** → Perfect Positive Correlation
- **0** → No Correlation
- **-1** → Perfect Negative Correlation

---

# Correlation Strength

| Correlation Value | Interpretation |
|-------------------|---------------|
| 0.00 – 0.19 | Very Weak |
| 0.20 – 0.39 | Weak |
| 0.40 – 0.59 | Moderate |
| 0.60 – 0.79 | Strong |
| 0.80 – 1.00 | Very Strong |

> The same ranges apply for negative values (e.g., -0.85 indicates a very strong negative correlation).

# Linear Regression

Linear Regression is a **Supervised Learning Regression Algorithm** that learns the relationship between **independent variables (features)** and a **continuous target variable** by fitting the **Best Fit Straight Line**.

```text
Data Points
      ●
   ●
      ●
         ●
            ●
-------------------------- Best Fit Line
```

---

# Equation of Linear Regression

```text
y = mx + c
```

Where

- **x** → Input Feature
- **y** → Predicted Output
- **m** → Slope (Weight)
- **c** → Intercept (Bias)

For multiple features

```text
y = w1x1 + w2x2 + ... + wnxn + b
```

---

# When to Use Linear Regression?

- Continuous target values.
- Linear relationship between features and target.
- Strong correlation.

Examples

- House Price Prediction
- Salary Prediction
- Sales Prediction

---

# When NOT to Use Linear Regression?

- Non-linear data.
- Weak correlation.
- Classification problems.

---

# Why Not for Non-linear Data?

Linear Regression fits only a **straight line**.

If the relationship is curved, predictions become inaccurate.

```text
Non-linear Data

      ●
   ●
 ●
     ●
         ●

Straight Line ❌
```

---

# What is Slope?

Slope represents the **rate of change of y with respect to x**.

Formula

```text
Slope (m) = Change in y / Change in x
```

or statistically

```text
m = Cov(X,Y) / Var(X)
```

---

## Example 1

| Study Hours | Marks |
|-------------|-------|
|1|10|
|2|20|
|3|30|
|4|40|

Here,

```
Slope = 10
```

Meaning

> Every additional hour of study increases marks by **10**.

---

# Covariance Formula

```text
Cov(X,Y)

=

Σ[(Xi - X̄)(Yi - Ȳ)] / n
```

---

# Variance Formula

```text
Var(X)

=

Σ(Xi - X̄)² / n
```

---

# Slope Formula

```text
Slope

=

Cov(X,Y)

/

Var(X)
```

---

# Intercept

Intercept is the predicted value of **y when x = 0**.

Formula

```text
Intercept (c)

=

Ȳ - mX̄
```

---

# Weight

Weight indicates the **importance of a feature**.

Example

```text
House Price

=

5000 × Area

+

200000
```

Here,

```
5000
```

is the weight.

Meaning:

Every additional square foot increases the house price by **₹5000**.

# Variance

Variance measures **how far data points are from the mean**.

- High Variance → Data is widely spread.
- Low Variance → Data is close to the mean.

If two datasets have similar variance, **Standard Deviation** is often used because it is easier to interpret (same unit as the data).

Scikit-learn's `LinearRegression` uses the **Ordinary Least Squares (OLS)** method.

OLS finds the line that minimizes the **sum of squared residuals (errors)**.

```text

      │

# Polynomial Features (Polynomial Regression)

Polynomial Features is a **feature engineering (preprocessing) technique** used to convert a linear feature into higher-degree features such as \(x^2\), \(x^3\), etc. This allows a **Linear Regression model** to learn **non-linear relationships** between the input and the target.

> **Important:** Polynomial Features **do not train the model**. They only create new features. The training is still done by **Linear Regression**.

---

# Why do we need Polynomial Features?

Sometimes the relationship between the feature and target is **not linear**.

For example,

```
Age  ---> Salary
```

may not always follow a straight line.

Instead, it may look like:

```
      *
    *   *
  *       *
 *         *
```

A simple Linear Regression model cannot fit this pattern properly.

Polynomial Features create additional features like **x²**, **x³**, etc., allowing Linear Regression to fit curved relationships.

---

# When should we use Polynomial Features?

- When the data shows a **curved (non-linear) relationship**.
- When Linear Regression gives poor performance due to underfitting.
- Only on **feature columns (X)**.
- Never apply Polynomial Features to the target column (y).

---

Example

```

It simply learns this transformation.

## 2. transform()

After learning the pattern,

transform() actually creates the new features.

Example

|---|
|2|
|3|
|4|

After

| X | X² |
|---|---|
|2|4|
|3|9|
|4|16|

# Why do we use fit_transform() on Training Data?

Training data is used to learn

- the degree
- feature pattern
- feature mapping

Therefore,

```python
updated_xtrain = poly.fit_transform(x_train)
```

# Step 3 : Transform Test Data

```python
updated_xtest = poly.transform(x_test)
```

Notice that we **do not use fit_transform()** here.

Only

```python
transform()
```

is used.

Why?

Because the transformation has already been learned from the training data.

The same transformation must be applied to the test data.

---

# Why NOT fit_transform() on Test Data?

Suppose

Training

```

Features created

x
x²
x³
```

The Linear Regression model is trained on **3 features**.

If we again use

```python
fit_transform(x_test)
```

the transformation is learned again, which may lead to inconsistencies and is considered incorrect practice.

Therefore,

Always use

```python
poly.transform(x_test)
```

# Advantages

- Captures non-linear relationships.
- Improves Linear Regression performance on curved data.
- Simple to implement.
- Works well for small and medium-sized datasets.

---

# Disadvantages

- Creates many additional features.
- High-degree polynomials may lead to overfitting.
- Increases computational cost.
- Harder to interpret as degree increases.

---

# Best Practices

- Apply Polynomial Features only to **feature columns (X)**.
- Perform **Train-Test Split** before applying transformations.
- Use `fit_transform()` only on training data.
- Use `transform()` only on testing/new data.
- Choose the polynomial degree carefully.
- Evaluate both Training and Testing scores to detect overfitting.

---

# Interview Questions

### Difference between fit(), transform(), and fit_transform()

| Method | Purpose |
|---------|---------|
| fit() | Learns the transformation pattern. |
| transform() | Applies the learned transformation. |
| fit_transform() | Performs fit() and transform() together. |

---

### Does PolynomialFeatures train the model?

**No.**

It is only a preprocessing (feature engineering) technique.

The model is trained by **LinearRegression**.

---

### Why use transform() on test data?

To ensure the test data has the **same number and structure of features** as the training data.

---

### Can Polynomial Regression work with Linear Regression?

**Yes.**

Polynomial Regression is simply **Linear Regression trained on polynomial features**.

# Easy Revision (One-Minute Summary)

- Polynomial Features is a **feature engineering technique**.
- Used when data has a **non-linear relationship**.
- Creates new features like **x², x³, x⁴**.
- `fit()` learns how to create new features.
- `transform()` creates those features.
- `fit_transform()` = fit() + transform().
- Use `fit_transform()` only on **training data**.
- Use `transform()` only on **testing/new data**.
- PolynomialFeatures **does not train the model**.
- Linear Regression performs the actual training.
- Always transform new data before prediction.
- Higher polynomial degrees can lead to **overfitting**.

## Other Evaluation pattern...

### 1) Underfit        
  - Model to too simple and unable to capture the complex pattern : 
  - Bias : High and Variance : Low which means (model will capture the simple pattern only)
  - Train And test data are low : Train(30%) and test(25%)
  - High Errors on both train and test data.
  - Example : Non - Linear Data - LinearRegression model

#### Solution for Underfit is use polynomial features for the non-linear data..
  - To make model good fit use polynomial Features or Regression
  - we can also use complex algorithm or model that can capture complex patterns like Random Forest etc.


### 2) Good Fit        
  - Model captures the complex pattern : 
  - Bias : Low and Variance : Low
  - Train And test data are high : Train(80%) and test(78%)
  - Low Errors on both train and test data.

  - Example : Linear Data - LinearRegression model

  - Generalization : It specifies, how the model performs on unseen data ,
  - if the model is not generalized well, it is overfitting and solution for that is PCA and Regularization.

### 3) Overfit        
  - Model to too complex and captures the noisy pattern :
  - Model captures the train data very well but fails to capture the pattern of test or unseen data.
  - Bias : Low and Variance :High :  which means (model will memorize the train data  pattern)
  - Train data is high and test data are low : Train(90%) and test(20%)
  - High Errors on test data Low errors on Train data..

  - Example : Non - Linear Data - LinearRegression model

   - MultiCollinearity : When the correlation between the feature columns is highly corrleated then that leads to problem called as multicollinearity..which leads to overfitting ..

#### Solution for overfit is:
   - PCA : Principal component Analysis
   - Regularization

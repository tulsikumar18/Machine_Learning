# Naive Bayes

## What is Naive Bayes?

**Naive Bayes** is a **Supervised Machine Learning Algorithm** that is mainly used for **Classification Problems**.

It predicts the class label using **Probability** and is based on **Bayes' Theorem**.

The model assigns the input to the class having the **highest probability**.

```text
               Input Data
                    │
                    ▼
          Calculate Probability
                    │
                    ▼
        Compare Class Probabilities
                    │
                    ▼
      Select Highest Probability
                    │
                    ▼
             Predicted Class
```

---

# Why is it called "Naive"?

The word **Naive** means **simple assumption**.

Naive Bayes assumes that **all features are independent of each other**, even though this assumption is usually not true in real-world datasets.

Example

Suppose we want to predict whether a person will buy a laptop.

Features

- Age
- Salary
- Occupation

Naive Bayes assumes

```text
Age ⟂ Salary ⟂ Occupation
```

Each feature contributes independently to the prediction.

---

# Bayes' Theorem

Naive Bayes is based on **Bayes' Theorem**.

## Formula

```text
             P(B|A) × P(A)
P(A|B) = -----------------------
               P(B)
```

or

```text
             P(X|Y) × P(Y)
P(Y|X) = -----------------------
                P(X)
```

Where

- **P(Y|X)** → Posterior Probability
- **P(X|Y)** → Likelihood
- **P(Y)** → Prior Probability
- **P(X)** → Evidence

---

# Understanding the Formula

Suppose we want to predict whether an email is Spam.

```text
P(Spam | Email)

↓

Probability that the email is Spam
given the email's words.
```

---

# Components of Bayes Theorem

## 1. Prior Probability

Probability of an event before seeing any data.

Example

```text
Out of 100 emails

20 are Spam

Prior Probability

P(Spam)

= 20/100

= 0.20
```

---

## 2. Likelihood

Probability of observing the given features if the class is known.

Example

```text
P("Lottery" | Spam)
```

Probability that the word **Lottery** appears in Spam emails.

---

## 3. Evidence

Overall probability of observing the feature.

Example

```text
P("Lottery")
```

Probability that the word **Lottery** appears in all emails.

---

## 4. Posterior Probability

Final probability after considering the evidence.

Example

```text
P(Spam | Lottery)
```

Probability that an email is Spam if it contains the word **Lottery**.

---

# Working of Naive Bayes

```text
Training Data
      │
      ▼
Calculate Prior Probability
      │
      ▼
Calculate Likelihood
      │
      ▼
Store Probabilities
      │
      ▼
New Data
      │
      ▼
Apply Bayes Theorem
      │
      ▼
Calculate Probability of Every Class
      │
      ▼
Choose Highest Probability
      │
      ▼
Prediction
```

# Types of Naive Bayes

There are **three major types**.

```text
              Naive Bayes
                   │
      ┌────────────┼─────────────┐
      │            │             │
      ▼            ▼             ▼
 Gaussian     Multinomial    Bernoulli
```

---

# 1. Gaussian Naive Bayes

Used when the input features are **Continuous Numerical Data**.

It assumes the data follows a **Normal (Gaussian) Distribution**.

Examples

- Height
- Weight
- Salary
- Age
- Temperature

```text
          Bell Curve

              ▲
             ▲ ▲
            ▲   ▲
          ▲       ▲
        ▲           ▲
──────▲──────────────▲──────
```

### Import

```python

```

---

# 2. Multinomial Naive Bayes

Used when features represent **Counts or Frequencies**.

Mostly used for **Text Classification**.

Examples

- Count Vectorizer
- TF-IDF
- Word Frequency

Applications

- Spam Detection
- Sentiment Analysis
- News Classification

### Import

```python

```

---

# 3. Bernoulli Naive Bayes

Used when input features contain only

- 0
- 1

(Binary Features)

Examples

```text
Contains Link?

No = 0
```

```text

No = 0
```

Mostly used after

- Binary Count Vectorizer
- Binary Features

Applications

- Spam Detection
- Email Classification

### Import


````python
# ============================================================
# Import Required Libraries
# ============================================================

import pandas as pd

# Import Multinomial Naive Bayes classifier
from sklearn.naive_bayes import MultinomialNB

# Split dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Pipeline helps combine preprocessing and model into one object
from sklearn.pipeline import Pipeline

# Used to evaluate model performance
from sklearn.metrics import classification_report

# Converts text into numerical features using TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
````

````python

# ============================================================
# Load Dataset
# ============================================================

# Read the IMDB movie review dataset
df = pd.read_csv("../datasets/IMDB Dataset.csv")

# Display first 5 rows
print(df.head())
````

````python


# ============================================================
# Separate Features and Target
# ============================================================

# Input Feature (Movie Reviews)
X = df["review"]

# Target Column (positive / negative)
y = df["sentiment"]

````

````python

# ============================================================
# Train-Test Split
# ============================================================

# 80% Training Data
# 20% Testing Data

xtrain, xtest, ytrain, ytest = train_test_split(
    X,
    y,
    train_size=0.80,
    random_state=42
)
````

````python


# ============================================================
# Create Machine Learning Pipeline
# ============================================================

# Step 1 : Convert text into TF-IDF vectors
# Step 2 : Train Multinomial Naive Bayes

main_pipeline = Pipeline(

    steps=[

        ("vectorizer", TfidfVectorizer()),

        ("model", MultinomialNB())

    ]

)

````

````python

# ============================================================
# Train the Model
# ============================================================

main_pipeline.fit(xtrain, ytrain)


# ============================================================
# Predictions
# ============================================================

# Prediction on Training Data
ytrain_pred = main_pipeline.predict(xtrain)

# Prediction on Testing Data
ytest_pred = main_pipeline.predict(xtest)
````

````python


# ============================================================
# Model Evaluation
# ============================================================

print("Training Classification Report")
print(classification_report(ytrain, ytrain_pred))

````

````python

print("Testing Classification Report")
print(classification_report(ytest, ytest_pred))



````

````python

# ============================================================
# Conclusion
# ============================================================

# If the training and testing accuracy are both high
# and their values are close to each other,
# then the model is considered a Good Fit.

# If training accuracy is much higher than testing accuracy,
# then the model is Overfitting.

# If both training and testing accuracies are low,
# then the model is Underfitting.
````

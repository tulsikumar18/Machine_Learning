# Imputation (Handling Missing Values)

## What is Imputation?

**Imputation** is a **Data Preprocessing Technique** used to handle **missing values (NaN/None)** in a dataset by replacing them with appropriate values instead of deleting the records.

Missing values can negatively affect the performance of Machine Learning models. Imputation helps preserve the dataset while making it suitable for model training.

```text
Dataset
   │
   ▼
Missing Values (NaN / None)
   │
   ▼
Imputation
   │
   ▼
Complete Dataset
   │
   ▼
Machine Learning Model
```

---

# Why do we need Imputation?

Many Machine Learning algorithms cannot handle missing values directly.

Without imputation:

- Model training may fail.
- Accuracy may decrease.
- Valuable data may be lost if rows are deleted.

Therefore, we replace missing values using appropriate strategies.

---

# SimpleImputer

**SimpleImputer** is a preprocessing technique provided by Scikit-Learn to replace missing values using different strategies.

Import

```python
from sklearn.impute import SimpleImputer
```

---

# Strategies in SimpleImputer

There are four commonly used strategies.

| Strategy | Used For | Replaces Missing Values With |
|-----------|----------|------------------------------|
| `mean` | Numerical Data | Mean of the column |
| `median` | Numerical Data (with outliers) | Median of the column |
| `most_frequent` | Numerical/Categorical Data | Most frequent value (Mode) |
| `constant` | Numerical/Categorical Data | User-defined value |

---

# 1. Mean Strategy

Used for **numerical columns without significant outliers**.

Formula

```text
Mean

=

Sum of values

------------------

Total number of values
```

Example

| Age |
|-----|
|20|
|25|
|30|
|NaN|

Mean

```text
(20 + 25 + 30)

---------------

3

=

25
```

Missing value becomes

```text
25
```

---

# 2. Median Strategy

Used for **numerical columns containing outliers**.

Formula

```text
Middle value after sorting
```

Example

```text
10

15

18

20

1000
```

Median

```text
18
```

Median is preferred because it is not affected much by outliers.

---

# 3. Most Frequent Strategy

Replaces missing values with the **most frequently occurring value (Mode)**.

Works for:

- Numerical Data
- Categorical Data

Example

| Gender |
|----------|
|Male|
|Female|
|Male|
|NaN|

Most Frequent

```text
Male
```

Missing value becomes

```text
Male
```

---

# 4. Constant Strategy

Replaces missing values with a user-defined value.

Examples

For numerical columns

```python
fill_value = 0
```

For categorical columns

```python
fill_value = "Unknown"
```

---

# Workflow of SimpleImputer

```text
Dataset
     │
     ▼
Find Missing Values
     │
     ▼
Choose Strategy
(Mean / Median / Mode / Constant)
     │
     ▼
Fit
(Learn replacement values)
     │
     ▼
Transform
(Replace missing values)
     │
     ▼
Complete Dataset
```

---

# fit() vs transform()

## fit()

Learns the replacement value for each column based on the selected strategy.

Example

```text
Age

↓

Mean = 35
```

Salary

```text
↓

Mean = 52000
```

The imputer stores these values internally.

---

## transform()

Uses the learned values to replace the missing values.

Example

```text
Age

NaN

↓

35
```

Salary

```text
NaN

↓

52000
```

---

## fit_transform()

Performs both operations together.

```text
Fit

↓

Learn

↓

Transform

↓

Replace Missing Values
```

      │                Mean
      │

---

# Advantages

- Very easy to use.
- Prevents loss of valuable data.
- Works well with pipelines.
- Supports numerical and categorical data.
- Faster than manually filling multiple columns.

---

# Disadvantages

- Mean can be affected by outliers.
- May introduce bias if many values are missing.
- Does not preserve relationships between features.

---

# SimpleImputer vs fillna()

| SimpleImputer | fillna() |
|---------------|----------|
| Scikit-Learn preprocessing tool | Pandas function |
| Works well with Pipelines | Cannot be directly used inside Scikit-Learn Pipelines |
| Learns values using `fit()` | Manually specify replacement |
| Better for ML workflows | Suitable for basic data cleaning |



# Interview Questions


### Why use SimpleImputer instead of fillna()?

Because it integrates with Scikit-Learn pipelines and learns replacement values using `fit()`.

---

### Does SimpleImputer accept 1D or 2D input?

It is generally used with **2D input (DataFrame or array)**.

````python
import pandas as pd

df = pd.DataFrame({
    'age' : [12,32,None,23,32,53,None,63,23,None],
    'sal' : [1300, None, 1400, None,1500,3500,7200, None,1200,1800],
    'gender' : ['F', None, 'M','M', 'F', 'F', 'M', 'F', None , None],
    'approved':[0,1,0,1, None,0,0,1,1,None]
})
````

````python
df.info()
````

````python
df.head()
````

````python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline


imputer = SimpleImputer(strategy='constant', fill_value='unknown')
imputer.fit_transform(df)
````

````python

imputer = SimpleImputer(strategy='mean')
imputer.fit_transform(df[['age','sal']]) ## it accepts 2D values..
````

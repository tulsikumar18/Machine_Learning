# K-Nearest Neighbors (KNN)

## What is KNN?

K-Nearest Neighbors (KNN) is a **Supervised Machine Learning Algorithm** used for both:

- Classification
- Regression

KNN is a **non-parametric** and **lazy learning algorithm** because it does not build a model during training. Instead, it stores the training data and performs computation only during prediction.

---

# Why is it called Lazy Learning?

- No model is trained.
- It simply stores all training data.
- Computation happens only when a new data point needs to be predicted.

```
Training
    │
    ▼
Store Training Data
    │
    ▼
Prediction Time
    │
    ▼
Calculate Distances
    │
    ▼
Find K Nearest Neighbors
    │
    ▼
Predict Output
```

# Working of KNN

Step 1:
Store all the training data.

↓

Step 2:
When a new data point arrives,
calculate the distance between the new point and every training sample.

↓

Step 3:
Sort all distances in ascending order.

↓

Step 4:
Select the first **K nearest neighbors**.

↓

Step 5:

- Classification → Majority Voting
- Regression → Average of K neighbors

# Distance Metrics

KNN predicts based on the nearest neighbors using distance formulas.

## 1. Euclidean Distance

Most commonly used.

\[
d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
\]

Example

```
Point A = (2,3)

Point B = (5,7)

Distance

= √((5-2)²+(7-3)²)

= √(9+16)

= √25

= 5
```

---

## 2. Manhattan Distance

Also called **City Block Distance**.

Used when movement is restricted to horizontal and vertical directions.

Example:
Google Maps (grid-like roads)

Formula

\[
d=|x_2-x_1|+|y_2-y_1|
\]

Example

```
Point A = (2,3)

Point B = (5,7)

Distance

= |5-2| + |7-3|

= 3 + 4

= 7
```

---

# Euclidean vs Manhattan

| Euclidean | Manhattan |
|-----------|-----------|
| Straight-line distance | Grid distance |
| Shortest path | Horizontal + Vertical path |
| Uses square root | Uses absolute values |
| Most commonly used | Used in grid-based movement |

# Choosing K Value

The value of **K** represents the number of nearest neighbors.

Example

```
K = 3

Take nearest 3 neighbors.
```

---

# Why should K be Odd?

For binary classification,

Choose an **odd value of K** to avoid ties.

Example

```
K = 6

YES = 3

NO = 3

Tie ❌
```

Instead,

```
K = 5

YES = 3

NO = 2

Prediction = YES ✅
```

---

# Rule to Select K

A common rule is

\[
K=\sqrt{N}
\]

Where

- N = Number of training samples

Example

```
Training Samples = 100

K = √100

K = 10

Usually choose the nearest odd value

K = 9 or 11
```

# Advantages

- Simple to understand
- Easy to implement
- No training phase
- Works for Classification and Regression
- Can learn complex decision boundaries

---

# Disadvantages

- Slow prediction on large datasets
- High memory usage
- Sensitive to irrelevant features
- Sensitive to outliers
- Performance decreases with high-dimensional data (Curse of Dimensionality)

---

# Interview Questions

### Why is KNN called a Lazy Learning Algorithm?

Because it does not build a model during training. It simply stores the training data and performs computation during prediction.

---

### Why should K be odd?

To avoid ties during binary classification.

---

### Why is Feature Scaling important in KNN?

KNN is distance-based. Features with larger values dominate the distance calculation.

Therefore, apply:

- StandardScaler
- MinMaxScaler

before training KNN.

---

### Can KNN perform Regression?

Yes.

- Classification → Majority Voting
- Regression → Average of nearest neighbors

````python
import pandas as pd
from sklearn.model_selection import GridSearchCV,train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder


````

````python
df = pd.read_csv('datasets/loan_data.csv')
df.head()
````

````python
df.tail()
````

````python
X = df.drop(columns='loan_status')
y = df.loan_status
````

````python
xtrain,xtest,ytrain,ytest = train_test_split(X,y, random_state=42, train_size=0.8)
````

## To check which scaling and encoding techniques we willl use , segregate the numerical and categorical column  and check for the unique values (nunique()), outliers (correlation matrix)

````python
nums_cols = X.select_dtypes(include='number').columns
obj_cols = X.select_dtypes(include='object').columns
````

````python
obj_cols.drop('person_education')
````

````python
order = ['High School','Bachelor','Master','Associate']
````

````python
preprocessing = ColumnTransformer(

    transformers=[

        (
            'onehot',
            OneHotEncoder(handle_unknown='ignore'),
            obj_cols.drop('person_education')
        ),

        (
            'ordinal',
            OrdinalEncoder(
                categories=[order],
                handle_unknown='use_encoded_value',
                unknown_value=-1
            ),
            ['person_education']
        )

    ],

    remainder='passthrough'
)

main_pipeline = Pipeline(

    steps=[

        ('preprocessing',preprocessing),

        ('model',DecisionTreeClassifier(random_state=42))

    ]
)

grid_search = GridSearchCV(

    estimator=main_pipeline,

    # we have to use model__ to access the parameter of model in main pipeline .
    # If we don't mention model__ it will not understand the parameter given here and 
    # it will give me Error 

    # Another approach we have is to directly mention model in estimator and then parameters in param_grid
    # IN this case , No need of mentioning the model__ 

    # estimator = 'DecisionTreeClassifier',
    # param_grid = { 'max_depth' ::[None,5,10,15,20], 'min_samples_leaf' :[1,2,5] }

    param_grid={

        'model__criterion':['gini','entropy'],  


        'model__max_depth':[None,5,10,15,20],

        'model__min_samples_split':[2,5,10],

        'model__min_samples_leaf':[1,2,5],

        'model__splitter':['best','random']

    },

    cv=5,

    scoring='accuracy',

    n_jobs=-1,

    verbose=2

)

grid_search.fit(xtrain,ytrain)
````

````python
# Best Parameters
print("Best Parameters:")
print(grid_search.best_params_)

# Best CV Score
print("Best Accuracy:")
print(grid_search.best_score_)

# Prediction
prediction = grid_search.predict(xtest)

# Evaluation
print(classification_report(ytest,prediction))
````

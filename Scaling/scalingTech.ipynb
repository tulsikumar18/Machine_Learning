# Feature Scaling

Feature Scaling is a **data preprocessing technique** used to bring numerical features to a similar scale. It ensures that features with larger values do not dominate features with smaller values during model training.

> **Example**

| Age | Salary |
|----:|-------:|
| 28 | 40000 |
| 25 | 30000 |
| 22 | 45000 |
| 21 | 6000 |
| 27 | 65000 |

Here, the **Age** feature ranges from **21–28**, whereas the **Salary** feature ranges from **6,000–65,000**. Since both features have different ranges, scaling is required.

---

# Why do we need Feature Scaling?

- Gives equal importance to all numerical features.
- Prevents large-valued features from dominating small-valued features.
- Improves model performance and prediction accuracy.
- Speeds up the training process for many machine learning algorithms.
- Helps distance-based algorithms calculate distances fairly.

---

# When should we use Feature Scaling?

- Apply scaling **only on numerical feature columns**.
- Use scaling when numerical features have different ranges.
- Scaling is applied **only on the feature columns (X)**, **not on the target column (y)**.
- Always perform **Train-Test Split before scaling** to avoid data leakage.

---

# Algorithms that require Scaling

Scaling is recommended for:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors (KNN)
- Support Vector Machine (SVM)
- K-Means Clustering
- Principal Component Analysis (PCA)
- Neural Networks
- Ridge Regression
- Lasso Regression
- ElasticNet Regression

### Tree-based algorithms generally do NOT require scaling

- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost

# Types of Feature Scaling

## 1. Standardization

Standardization scales the data so that:

- **Mean = 0**
- **Standard Deviation = 1**

The scaled values **do not have a fixed range**. They can be positive or negative.

---

### Types of Standardization

## A. Standard Scaler

Standard Scaler uses the **Mean** and **Standard Deviation** to scale the data.

### Formula

```text
z = (x - μ) / σ
```

Where:

- **x** = Original value
- **μ (Mu)** = Mean of the feature
- **σ (Sigma)** = Standard Deviation

### Example

| Original Salary | Scaled Value |
|---------------:|-------------:|
| 6000 | -1.42 |
| 30000 | -0.35 |
| 40000 | 0.12 |
| 45000 | 0.34 |
| 65000 | 1.31 |

### Advantages

- Simple and widely used.
- Works well when the data is approximately normally distributed.
- Improves the performance of many machine learning algorithms.

### Limitations

- Sensitive to outliers because it uses the **Mean**, which is affected by extreme values.

### Best Used When

- Data has **few or no outliers**.
- Data is approximately normally distributed.

---

## B. Robust Scaler

Robust Scaler uses the **Median** and **Interquartile Range (IQR)** instead of Mean and Standard Deviation.

Since the **Median** is less affected by outliers, Robust Scaler performs well on datasets containing many outliers.

### Formula

```text
Scaled Value = (x - Median) / IQR
```

Where:

- **Median** = Middle value
- **IQR (Interquartile Range)** = Q3 - Q1

### Advantages

- Not affected much by outliers.
- Produces more reliable scaling for skewed datasets.

### Limitations

- Does not make the mean equal to 0 or the standard deviation equal to 1.

### Best Used When

- Dataset contains **many outliers**.

---

# 2. Normalization

Normalization scales all values into a **fixed range between 0 and 1**.

The most commonly used normalization technique is **Min-Max Scaling**.

### Formula

```text
Scaled Value = (x - Min) / (Max - Min)
```

Where:

- **Min** = Minimum value of the feature
- **Max** = Maximum value of the feature

### Example

Original Age:

```
21
22
25
27
28
```

After Min-Max Scaling

```
0.00
0.14
0.57
0.86
1.00
```

### Advantages

- Values always lie between **0 and 1**.
- Easy to understand and interpret.
- Useful for algorithms that require bounded input values.

### Limitations

- Highly sensitive to outliers because it depends on the minimum and maximum values.

### Best Used When

- Dataset has **no significant outliers**.
- Algorithms require values within a fixed range.

---

# Standardization vs Normalization

| Feature | Standardization | Normalization |
|----------|-----------------|---------------|
| Output Range | No fixed range | 0 to 1 |
| Formula | `(x - μ) / σ` | `(x - Min) / (Max - Min)` |
| Uses Mean | ✅ Yes | ❌ No |
| Uses Standard Deviation | ✅ Yes | ❌ No |
| Uses Min & Max | ❌ No | ✅ Yes |
| Sensitive to Outliers | Standard Scaler: Yes<br>Robust Scaler: No | Yes |
| Best For | Normally distributed data | Fixed-range scaling |

---

# Summary

| Scaler | Uses | Best When |
|---------|------|-----------|
| **Standard Scaler** | Mean & Standard Deviation | Few or no outliers |
| **Robust Scaler** | Median & IQR | Many outliers |
| **Min-Max Scaler** | Minimum & Maximum | No significant outliers and values need to be between 0 and 1 |

---

# Interview Points

- Feature Scaling is applied **only on numerical features**.
- Always perform **Train-Test Split before scaling**.
- Fit the scaler **only on the training data**, then transform both training and testing data.
- Scaling improves the performance of **distance-based** and **gradient-based** algorithms.
- Tree-based algorithms generally **do not require scaling**.
- **Standard Scaler** uses Mean and Standard Deviation.
- **Robust Scaler** uses Median and IQR.
- **Min-Max Scaler** scales values between **0 and 1**.
- Scaling changes only the magnitude of the values; it does **not** change the relationship between data points.

# Easy Trick to Remember

### Standard Scaler

- Uses **Mean**
- Few Outliers
- Mean = 0
- Standard Deviation = 1

### Robust Scaler

- Uses **Median**
- Many Outliers
- Resistant to Extreme Values

### Min-Max Scaler

- Uses **Minimum and Maximum**
- Output Range = **0 to 1**
- Best when there are no significant outliers
